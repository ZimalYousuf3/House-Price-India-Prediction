```python
# Handling Columns: dropping columns

df = df.drop(columns=['id'])
```

```python
# Train/Test split

from sklearn.model_selection import train_test_split

x = df.drop(columns=['Price'])
y = df['Price']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(f'Training set size: {len(x_train)}')
print(f'Testing set size: {len(x_test)}')
```

```python
# Linear Regression: Predicting

from sklearn.linear_model import LinearRegression

lin = LinearRegression()
lin.fit(x_train, y_train)
lin.preds = lin.predict(x_test)
```

```python
# Linear Regression: Metrics

from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score

MAE = mean_absolute_error(lin.preds, y_test)
RMSE = root_mean_squared_error(lin.preds, y_test)
R2 = r2_score(lin.preds, y_test)

print(f'Linear Regression:-')
print(f'MAE: {MAE}')
print(f'RMSE: {RMSE}')
print(f'R2: {R2}')
```

```python
# Random Forest: Predicting

from sklearn.ensemble import RandomForestRegressor

forest = RandomForestRegressor(n_estimators=100, random_state=42)
forest.fit(x_train, y_train)
forest.preds = forest.predict(x_test)
```

```python
# Random Forest: Metrics

mae = mean_absolute_error(forest.preds, y_test)
rmse = root_mean_squared_error(forest.preds, y_test)
r2 = r2_score(forest.preds, y_test)

print(f'Random Forest:-')
print(f'MAE: {mae}')
print(f'RMSE: {rmse}')
print(f'R2: {r2}')
```

```python
# Importances

importances = pd.Series(forest.feature_importances_, index=x.columns)
importance = importances.sort_values(ascending=False)

print(importance)
```
